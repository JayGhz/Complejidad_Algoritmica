Implemente un algoritmo para gestionar el grupo de amigos descrito en la primera parte.

```python
class UnionFind:
    def __init__(self, n):
        self.parent = list(range(n)) # Cada nodo es su propio padre

    def find(self, i):
        # Encuentra el padre de i
        if self.parent[i] == i:
            return i
        else:
            self.parent[i] = self.find(self.parent[i])
            return self.parent[i]
    
    def union(self, i, j):
        # Une los conjuntos de i y j
        root_i = self.find(i)
        root_j = self.find(j)

        if root_i == root_j: # Si ya estan en el mismo conjunto
            return
        else:
            self.parent[root_i] = root_j
    
    def are_friends(self, i, j):
        return self.find(i) == self.find(j)
```

```python
individuos = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j']
uf = UnionFind(len(individuos)) # Array de 10 individuos
print(uf.parent)
```

```python
# Unimos a y b
uf.union(0, 1)
print(uf.parent)
```

```python
# Algoritmo para gestionar el grupo de amigos
def grupo_amigos(uf, individuos):
    amigos = {}
    for i in range(len(individuos)):
        root = uf.find(i)
        if root not in amigos:
            amigos[root] = []
        amigos[root].append(individuos[i])
    grupo_n = 1
    for grupo in amigos.values():
        print(f'Grupo {grupo_n}: {grupo}')
        grupo_n += 1

grupo_amigos(uf, individuos)
```

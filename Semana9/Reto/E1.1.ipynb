#### Ejercicio  1.1  
Implemente  las  funciones  Find y  Union.

```python
class UnionFind:
    def __init__(self, n):
        self.parent = list(range(n)) # Cada nodo es su propio padre

    def find(self, i):
        # Encuentra el padre de i
        if self.parent[i] == i:
            return i
        else:
            self.parent[i] = self.find(self.parent[i])
            return self.parent[i]
    
    def union(self, i, j):
        # Une los conjuntos de i y j
        root_i = self.find(i)
        root_j = self.find(j)

        if root_i == root_j: # Si ya estan en el mismo conjunto
            return
        else:
            self.parent[root_i] = root_j
    
    def are_friends(self, i, j):
        return self.find(i) == self.find(j)
```

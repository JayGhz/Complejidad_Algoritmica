```python
#Generar un polinomio aleatorio de grado n y resolverlo asignado un valor a x
import random

def polinomio(n):
    coef = []
    for i in range(n+1):
        coef.append(random.randint(1,10))
    return coef

def evaluar(coef,x):
    pol = 0
    for i in range(len(coef)):
        pol += coef[i]*x**i
    return pol

def main():
    n = int(input("Ingrese el grado del polinomio: "))
    x = int(input("Ingrese el valor de x: "))
    coef = polinomio(n)
    
    print("El polinomio generado es: ")
    for i in range(len(coef)):
        if i == 0:
            print(coef[i],end="")
        else:
            print(" + ",coef[i],"x^",i,end="")

    print("\nEl valor del polinomio evaluado en " + str(x) + " es: ",evaluar(coef,x))


if __name__ == "__main__":
    main()




```

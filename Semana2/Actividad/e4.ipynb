#### Problema del pastor(acertijo del lobo, la cabra y la col) 
El problema consiste en encontrar todas las posibles combinaciones
para que el pastor pueda cruzar el rio sin que la cabra se coma la col
o el lobo se coma la cabra. Se simulare la posicion de cada uno de ellos
con una tupla de 4 elementos:

(0,0,0,0) -> (pastor, lobo, cabra, col) en la orilla inicial

(1,1,1,1) -> (pastor, lobo, cabra, col) en la orilla final

```python
# Explorar todas las posibles soluciones con fuerza bruta
def is_secure(state):
    pastor, wolf, cabra, col = state
    if wolf == cabra and pastor != wolf: # El lobo se come a la cabra
        return False
    if cabra == col and pastor != cabra: # La cabra se come al repollo
        return False
    return True
```

```python
def brute_force():
    states = [(p, w, c, co) for p in [0, 1] for w in [0, 1] for c in [0, 1] for co in [0, 1]]
    solution = []

    def find_solution(current_state, path):
        if current_state == (1, 1, 1, 1): 
            return path + [current_state]
        
        for p, l, c, co in states:
            new_state = (p, l, c, co)
            if is_secure(new_state) and new_state not in path:
                new_path = find_solution(new_state, path + [new_state])
                if new_path:
                    return new_path
        return None
    
    # Estado inicial del problema (todos en la orilla inicial)
    initial_state = (0, 0, 0, 0)
    solution = find_solution(initial_state, [])

    if solution:
        print("Solución encontrada:")
        print("\t(P, L, C, Co)")
        for path in solution:
            print("\t",path)
    else:
        print("No se encontró solución")
```

```python
#Encontrar un elemento en una lista de forma lineal
def lineal(A, n, key):
    for i in range(n):
        if A[i] == key:
            return "El elemento " + str(key) + " se encuentra en la posicion: " + str(i)
    return "El elemento " + str(key) + " no se encuentra en la lista"

numero = int(input("Ingrese el numero a buscar: "))
lineal([1, 10, 3, 8, 5], 5, numero) 
```

```python
def encontrar_ciclos(grafo):
    def recorrido(v, visitado, pila, ciclos):
        visitado[v] = True
        pila.append(v)

        for vecino in grafo[v]:
            if not visitado[vecino]:  # si el vecino no ha sido visitado
                recorrido(vecino, visitado, pila, ciclos)
            elif vecino in pila:  # si el vecino está en la pila, hay un ciclo
                ciclo = pila[pila.index(vecino):]  # extraer el ciclo desde el vecino
                ciclos.append(ciclo)

        pila.pop()  # eliminar el nodo de la pila al regresar

    n = len(grafo)
    visitado = [False] * n
    pila = []
    ciclos = []  

    for i in range(n):
        if not visitado[i]:
            recorrido(i, visitado, pila, ciclos)

    if ciclos:
        print("Ciclos encontrados:\n")
        for ciclo in ciclos:
            print(ciclo)
    else:
        print("No existe un ciclo en el grafo")

```

```python
grafo = {
    0: [1],
    1: [2],
    2: [0],  # Ciclo 1
    3: [4],
    4: [5],
    5: [3],  # Ciclo 2
    6: [7],
    7: []
}

encontrar_ciclos(grafo)
```

```python

grafo_sin_ciclos = [
    [1, 2],   # 0 -> 1, 2
    [3],      # 1 -> 3
    [3],      # 2 -> 3
    []        
]

encontrar_ciclos(grafo_sin_ciclos)
```

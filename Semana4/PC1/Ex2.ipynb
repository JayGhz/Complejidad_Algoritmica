##### **Tecnica usada: Divide y venceras** 
Dado que como pararametro de la funcion es una lista que contiene otras sublistas y se desea encontrar la sublista que cumpla la condicion planteada, use la tecnica de divide y venceras ya que facilita la busqueda de dicha sublista ya que la funcion divide la lista en sublistas(izquierda y derecha) recursivamente y compara ambas para encontrar la region o sublista con mayor poblacion usando la funcion sum( ). 

```python
def region_mayor_poblacion(poblacion: list):
    def region_mayor_poblacion_aux(poblacion: list, inicio: int, fin: int):
        if inicio == fin: # Caso donde solo hay una region
            return inicio
        else:
            mitad = (inicio + fin) // 2 # Dividir la lista en 2 mitades
            # Llamar recursivamente a la funcion para cada mitad
            region_izq = region_mayor_poblacion_aux(poblacion, inicio, mitad)
            region_der = region_mayor_poblacion_aux(poblacion, mitad + 1, fin)

            # Comparar la suma de la poblacion de las regiones de la izquierda y derecha
            if sum(poblacion[region_izq]) > sum(poblacion[region_der]): 
                return region_izq
            else:
                return region_der
    
    region = region_mayor_poblacion_aux(poblacion, 0, len(poblacion) - 1)

    print(f"Region con mayor poblacion: {poblacion[region]}")
    print(f"Poblacion: {sum(poblacion[region])}")
```

```python
poblacion = [[417, 346, 420, 432], [294, 308, 371, 452], [393, 333, 108, 246], [376, 136, 213, 174], [434, 422, 247, 307], [314, 272, 204, 293], [480, 186, 304, 223], [436, 426, 444, 366], [108, 150, 495, 418], [395, 491, 307, 422]]
region_mayor_poblacion(poblacion)
```

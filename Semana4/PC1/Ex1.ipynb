```python
def tot_palabras_por_linea(texto):
    if not texto: # Si no hay texto, retorna una lista vacia
        return [] 
    else:
        indice_salto_linea = texto.find("\n") # Encontar el indice del salto de linea
        if indice_salto_linea == -1: # Si no hay salto de linea (-1)
            return [len(texto.split())] # Retornar la cantidad de palabras de la linea actual usando split
        else:
            #Recursion para contar las palabras de la linea actual y las siguientes
            return [len(texto[:indice_salto_linea].split())] + tot_palabras_por_linea(texto[indice_salto_linea+1:])
```

```python
txt="Lorem ipsum dolor sit amet, consectetur adipiscing elit.\n Nam arcu arcu, euismod mollis rhoncus quis, egestas at leo.\n Maecenas luctus neque nulla, vel imperdiet lacus ultrices vel."
print(tot_palabras_por_linea(txt))
```

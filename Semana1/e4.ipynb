```python
#Escribir un progrma para calcular la distancia entre dos puntos en un plano cartesiano.


def distancia(p1, p2):
    return ((p1[0]-p2[0])**2 + (p1[1]-p2[1])**2)**0.5

p1 = input("Ingrese las coordenadas del primer punto (x, y): ")
p2 = input("Ingrese las coordenadas del segundo punto (x, y): ")

p1 = tuple(map(float, p1.split(',')))
p2 = tuple(map(float, p2.split(',')))

print("La distancia entre los puntos es:", distancia(p1, p2))
```

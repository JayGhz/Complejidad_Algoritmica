```python
#Escribir un programa para sumar dos numerosy pedir al usuario que ingrese dos numeros

def suma(a,b):
    a = int(input("Ingrese el primer numero: "))
    b = int(input("Ingrese el segundo numero: "))
    return a + b
```

```python
#Imprimir resultado
mensaje = "La suma de los dos numeros es: " + str(suma(0,0))
print(mensaje)
```

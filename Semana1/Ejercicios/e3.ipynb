```python
#Funcion para validar si la fila del sudoku es correcta
def validar_fila(sudoku, fila):
    return len(set(sudoku[fila])) == 9

#Funcion para validar si la columna del sudoku es correcta
def validar_columna(sudoku, columna):
    columna = [sudoku[fila][columna] for fila in range(9)]
    return len(set(columna)) == 9

#Funcion para validar si el cuadrante del sudoku es correcto
def validar_cuadrante(sudoku, fila, columna):
    cuadrante = []
    for i in range(3):
        for j in range(3):
            cuadrante.append(sudoku[fila+i][columna+j])
    return len(set(cuadrante)) == 9

#Funcion para validar si el sudoku esta correcto
def validar_sudoku(sudoku):
    for i in range(9):
        if not validar_fila(sudoku, i) or not validar_columna(sudoku, i):
            return False
    for i in range(0, 9, 3):
        for j in range(0, 9, 3):
            if not validar_cuadrante(sudoku, i, j):
                return False
    return True
```

```python
sudoku = [
    [5, 3, 4, 6, 7, 8, 9, 1, 2],
    [6, 7, 2, 1, 9, 5, 3, 4, 8],
    [1, 9, 8, 3, 4, 2, 5, 6, 7],
    [8, 5, 9, 7, 6, 1, 4, 2, 3],
    [4, 2, 6, 8, 5, 3, 7, 9, 1],
    [7, 1, 3, 9, 2, 4, 8, 5, 6],
    [9, 6, 1, 5, 3, 7, 2, 8, 4],
    [2, 8, 7, 4, 1, 9, 6, 3, 5],
    [3, 4, 5, 2, 8, 6, 1, 7, 9]
]

if validar_sudoku(sudoku):
    print("El sudoku es correcto")
else:
    print("El sudoku es incorrecto")
```

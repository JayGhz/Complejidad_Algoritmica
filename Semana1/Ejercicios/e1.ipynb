```python
#Funcion que genera un arreglo de tamaño n con numeros aleatorios
import random

def generate_array(n):
    arr = []
    for i in range(n):
        arr.append(random.randint(0,100))
    return arr
```

```python
#Mostrar arreglo
def show_array(arr):
    print(arr)
    
arr = generate_array(10)
show_array(arr)
```

```python
#Mostrar arreglo al reves
def reverse_array(arr):
    return arr[::-1]

arr = generate_array(10)

show_array(arr)
show_array(reverse_array(arr))
```

```python
def get_min(arr):
    min = arr[0]
    for i in arr:
        if i < min:
            min = i
    return min

arr = generate_array(10)
show_array(arr)

print('\nEl elemento minimo del arreglo es: ', get_min(arr))
```

```python
#Promedio de un arreglo
def average(arr):
    sum = 0
    for i in arr:
        sum += i
    return sum / len(arr)

arr = generate_array(10)
show_array(arr)
print('\nEl promedio del arreglo es: ', average(arr))
```

```python
#Ocurrencias de un numero en un arreglo
def ocurrences(arr):
    max_value = max(arr) if arr else None
    #Creamos un arreglo de tamaño max_value + 1
    s = [0] * (max_value + 1)

    #Contamos las ocurrencias
    for i in arr:
        s[i] += 1
    
    return s

arr = [1, 3, 2, 1, 4, 2, 1]
show_array(arr)
print('\nEl arreglo generado segun las ocurrencias de cada valor de i es: ', ocurrences(arr))
```

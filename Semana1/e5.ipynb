```python
import time

def factorial(n):
    repuesta = 1

    while n > 1:
        repuesta *= n
        n -= 1
    return repuesta


def factorial_r(n):
    if n == 1:
        return 1
    return n * factorial_r(n - 1)


if __name__ == '__main__':
    n = 2000
    start = time.time()
    factorial(n)
    final = time.time()
    print(final - start)

    start = time.time()
    factorial_r(n)
    final = time.time()
    print(final - start)
```

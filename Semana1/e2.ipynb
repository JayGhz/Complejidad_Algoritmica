```python
#Funcion que devuelve el mayor numero de dos numeros
def mayor(a,b):
    if a>b:
        return a
    else:
        return b
```

```python
a = int(input("Introduce el primer numero: "))
b = int(input("Introduce el segundo numero: "))

print("El mayor numero es: ",mayor(a,b))
```

```python
def mayor(a,b,c):
    if a > b and a > c:
        return a
    elif b > a and b > c:
        return b
    else:
        return c
    
```

```python
a = int(input("Introduce el primer numero: "))
b = int(input("Introduce el segundo numero: "))
c = int(input("Introduce el tercer numero: "))

print("El mayor entre ", a, b , c, "es: ", mayor(a,b,c))
```

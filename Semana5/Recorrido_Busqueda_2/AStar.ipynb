**Funcion BFS para recorrer una matriz de adyacencia**

```python
import numpy as np
import graphviz as gv
```

```python
#Funcion de recorrido en amplitud
def bfs(G, s):
    n = len(G)
    visited = [False]*n #Inicializamos todos los nodos como no visitados
    path = [-1]*n #Inicializamos el camino recorrido
    queue = [s] #Inicializamos la cola con el nodo de inicio
    visited[s] = True #Marcamos el nodo como visitado

    while queue:
        u = queue.pop(0)
        for v in range(n):
            if G[u][v] and not visited[v]:
                visited[v] = True
                path[v] = u
                queue.append(v)
    return path
```

```python
#Funcion para dibujar el grafo usando graphviz, indicando si es dirigido o no
#Colorear la ruta resultante del recorrido en amplitud

def draw_graph(G, directed=False, path=None):
    n = len(G)
    dot = gv.Digraph() if directed else gv.Graph()
    for u in range(n):
        dot.node(str(u))
    for u in range(n):
        for v in range(u+1, n):
            if G[u][v]:
                dot.edge(str(u), str(v))
    if path:
        for v in range(1, n):
            if path[v] != -1:
                dot.edge(str(path[v]), str(v), color='red')
    return dot
```

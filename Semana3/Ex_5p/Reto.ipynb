```python
#Validar la posicion para colocar un numero
def validar(arr, num, pos):

    ultima_pos = -1
    for i in range(len(arr)):
        if arr[i] == num:
            ultima_pos = i
    
    # si el numero apaarece menos de una vez, se puede colocar
    if ultima_pos < 1:
        return True
    
    # validar si la distancia entre la ultima posicion y la posicion actual es el mismo numero
    return pos - ultima_pos[-1] == num
```

```python
def combinaciones(arr, n, pos, sol):

    if pos == len(arr):
        sol.append(arr.copy())
        return
    
    if arr[pos] != 0: # si la posicion ya tiene un numero, pasa al siguiente
        combinaciones(arr, n, pos + 1, sol)
        return

    for num in range(1, n + 1):
        if arr.count(num) < 2 and validar(arr, num, pos):
            arr[pos] = num
            sgt_pos = pos + num + 1

            if sgt_pos < len(arr) and arr[sgt_pos] == 0:
                arr[sgt_pos] = num
                combinaciones(arr, n, pos + 1, sol)
                arr[sgt_pos] = 0
        
        arr[pos] = 0
```

```python
def generar_combinaciones(n):
    arr = [0] * (2 * n)
    sol = []
    combinaciones(arr, n, 0, sol)
    
    if sol:
        for s in sol:
            print(s)
    else:
        print("No hay solucion")

generar_combinaciones(7)
```
